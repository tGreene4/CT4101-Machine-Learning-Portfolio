## Missing Values
After consideration of the source material on which the dataset was developed, I find it safe to assume that the unrecorded values for stalk-root are derived from the lack of specific descriptor assigned to the stalk root in the original literature, allowing for the reassignment of NaN values as a separate class of stalk roots with a consistent similar lack of descriptors or notable properties, which I have denoted generic("g"). A simple imputer to replace the missing values will be the first step in the model pipeline.

```python
pipeline = make_pipeline(SimpleImputer(strategy="constant",fill_value="g"))
```

## Categorical Data Encoding
For Decision Trees as part of a Random Forest, Ordinal Encoding can be used as due to how Decision trees derive decision boundaries from information gain or Gini gain measures without dependence on absolute values.
