## Baseline
The baseline model proved the high effectiveness of the Random Forest model for the purposes of our dataset, and sets the standard against which we measure the effects of performance as features are removed:

```python
from sklearn.metrics import average_precision_score, f1_score
print(f"Accuracy: {baseline_pipeline.score(x_test,y_test)}")
test_predict = baseline_pipeline.predict(x_test)
print(f"Average Precision Score: {average_precision_score(y_test=='e',test_predict=='e')}")
print(f"F1 Score: {f1_score(y_test=='e',test_predict=='e')}")
```

## Specific Feature Culling

For all targeted metrics, the base model trained on any of the datasets generated with a culled feature performed with full accuracy on the test set, equivalent to the base model trained on all features. 

```python
#specific feature cull evaluation
specific_feature_result = {}
for feature in feature_list:
    baseline_pipeline.fit(specific_feature_removed[feature],y_train)
    test_predict = baseline_pipeline.predict(x_test)
    result = (baseline_pipeline.score(x_test,y_test),
              average_precision_score(y_test=='e',test_predict=='e'),
              f1_score(y_test=='e',test_predict=='e')
    )
    specific_feature_result[feature] = result
```

## Candidate Feature Pool Culling
Similarly, the decrease in the size of the pool of selected candidate features for each tree also did not impact the overall results of the model's predictions for the test set, which maintained 100% accuracy.

```python
feature_pool_result = []
for model in feature_pool_culled:
    model.fit(x_train,y_train)
    model.predict(x_test)
    result = (model.score(x_test,y_test),
                  average_precision_score(y_test=='e',test_predict=='e'),
                  f1_score(y_test=='e',test_predict=='e')
        )
    feature_pool_result.append(result)
```
