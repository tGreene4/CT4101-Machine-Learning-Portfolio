# Random Forest Classification of Mushrooms with Feature Culling
## Random Forest models for the Binary Classification of Edible and Poisonous Mushrooms and the Effects of Feature Removal

This project intends to examine the effects of removing both specific features from the overall dataset and candidate features from the randomized Random Forest feature split, as a test of the predictive power of specific features and the robustness of this category of model against loss of data or the capacity to acquire data on certain features.

As a testing problem, the model is being developed for the binary classification of mushrooms as edible or poisonous based on a variety of physical characteristics, a resulting in a wide feature set from which progressively smaller specified and randomized subsets can be sampled in order to test priority metrics in this domain context, namely variance and precision as protection against false positives.
