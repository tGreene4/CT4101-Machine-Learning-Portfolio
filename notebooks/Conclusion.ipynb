# Conclusion
## Model Analysis
In contrast to my initial expectations of the degredation of performance as data was removed, in the case of this dataset all Random Forests regardless of their tree feature subset sizes and removed features displayed consistently equivalent performance to the 100% accuracy of the base model. Under the current parameters none of the models expressed any degredation in performance.
* All models display 100% accuracy, and thus 100% and a 1.0 F1 Score
* Models of feature subset sizes (22,1) and models trained on data with each individual feature removed all predicted at the same accuracy.
## Limitations
* Feature culling tests only cull one feature, complete removal of multiple features from the pool may yield further results but analysis of C(22,n) possibilities for n>1 was considered infeasible.
* Dataset is insufficiently complex: the dataset displays a simple correlation between single features and the target that make development of decision trees for the dataset relatively simple due to the high information gain of these features, resulting in a lack of complexity in these models that makes analysis of their performance degredation under simple dimension reduction more difficult and less conclusive.
## Next Steps
* Source a more complex dataset with an equivalent classification problem which involves more complex interdependence of features and varied samples to engage the model with a problem which will result in a greater error rate by which to measure precision and recall impacts.
* Widen the number of feature culling, candidate pool and forest size configurations.
