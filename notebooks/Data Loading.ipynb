My choice of dataset is "Mushroom", provided by the University of California, Irvine. It consists of roughly 8000 samples of mushrooms of two different families as compiled based on the Audubon Society Field Guide to North American Mushrooms(see Data Audit for more information).

For data loading, UC Irvine helpfully provide a packaged wrapper around the API for the UCI machine learning repository, allowing the dataset to be imported directly. "Mushrooms" is archived with id 73.

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
mushroom = fetch_ucirepo(id=73)
dataset = mushroom.data
```

The API also returns various metadata and dataset feature properties:

```python
print(mushroom.metadata,"\n")
print(mushroom.variables)
```
