# Dataset Audit
The Mushroom dataset is a compilation of 8124 hypothetical examples of mushrooms of 23 different species across the Agaricaceae family, primarily of the Agaricus and Lepiota genera, compiled based on species descriptions from the [Audubon Society Field Guide to North American Mushrooms(1981)](https://archive.org/details/audubonsocietyfi0000linc/page/500/mode/2up). The dataset contains 22 features, including habitat biome data, population density/growth type, and the colours, textures, patterning and other properties of various parts of the mushroom such as the gills, cap and stalk. Our target feature is "poisonous", which categorises mushrooms as either edible(e) or poisonous/of unknown edibility(p)
